```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib; matplotlib.use('Agg')
df = pd.read_csv(r"C:\Users\niluc\Downloads\archive\creditcard.csv")
print('Missing:', df.isnull() .sum().sum())
print(df.shape)
print(df['Class'].value_counts()) 
print(f"Fraud Rate: {df['Class'].mean() * 100:.4f}%")

#feature engineering
df['Log_amount'] = np.log1p(df['Amount'])
df['Hour'] = (df['Time'] // 3600) % 24
df['OffHours'] = ((df['Hour'] >=0) & (df['Hour'] <=5)).astype(int)
df.drop(['Time', 'Amount'], axis=1, inplace=True)

fig, axes = plt.subplots (2,2, figsize=(12,8))

#Class imbalance
counts =df['Class'].value_counts()
axes[0,0].bar(['Legitimate', 'Fraud'], counts.values,
              color=['#1B8CA6', '#C0392B'], edgecolor='white')
axes[0,0].set_title('Class Distribution(Extreme Imbalance)')
axes[0,0].set_ylabel('Count')
for i,(v) in enumerate(counts.values):
    axes[0,0].text(i, v+500, f'{v:,}', ha='center', fontsize=9)

#transaction amount by class
df.groupby('Class')['Log_amount'].plot (kind='kde', ax=axes[0,1], legend=True)
axes[0,1].set_title('Log(amount) distribution by Class')
axes[0,1].legend(['Legitimate','Fraud'])

#Fraud by hour
fraud_by_hour =df[df['Class']==1].groupby('Hour').size()
fraud_by_hour = fraud_by_hour.reindex(range(24), fill_value=0)
fraud_by_hour.plot (kind='bar',ax=axes[1,0], color='#C0392B', edgecolor='white')
axes[1,0].set_title('Fraud Transaction by Hour')
axes[1,0].set_xlabel('Hour of Day')
axes[1,0].tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.savefig('fraud_eda_charts.png', dpi=120)
df.to_csv('fraud_clean.csv',index=False)
print('EDA saved. Clean data saved.')



```
